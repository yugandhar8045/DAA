# Implement prim’s algorithm


```python

# Prim's Algorithm

def prim(graph, start):
    n = len(graph)
    visited = [False] * n
    visited[start] = True

    mst_cost = 0
    print("Minimum Spanning Tree:")

    for _ in range(n - 1):
        min_cost = float('inf')
        u = v = -1

        for i in range(n):
            if visited[i]:
                for j in range(n):
                    if not visited[j] and graph[i][j] != 0:
                        if graph[i][j] < min_cost:
                            min_cost = graph[i][j]
                            u, v = i, j

        if v == -1:
            print("Graph is not connected")
            return

        print(u, "-", v, ":", min_cost)
        mst_cost += min_cost
        visited[v] = True

    print("Minimum Cost:", mst_cost)


# Graph representation using adjacency matrix
graph = [
    [0, 2, 0, 6, 0],
    [2, 0, 3, 8, 5],
    [0, 3, 0, 0, 7],
    [6, 8, 0, 0, 9],
    [0, 5, 7, 9, 0]
]

prim(graph, 0)
```
